```python
!pip install yfinance -q
import yfinance as yf
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# 1. Get stock data (change AAPL to any stock, e.g. TCS.NS)
df = yf.download("AAPL", start="2018-01-01", auto_adjust=True)
price = df["Close"].squeeze()

# 2. TREND: price with moving averages
plt.figure(figsize=(12,5))
plt.plot(price, label="Price")
plt.plot(price.rolling(50).mean(), label="50-day average")
plt.plot(price.rolling(200).mean(), label="200-day average")
plt.title("Trend"); plt.legend(); plt.show()

# 3. SEASONALITY: break monthly prices into trend + repeating pattern
monthly = price.resample("ME").mean()
seasonal_decompose(monthly, model="multiplicative", period=12).plot()
plt.show()

# 4. Average return per month
ret = price.resample("ME").last().pct_change() * 100
ret.groupby(ret.index.month).mean().plot(kind="bar", title="Average return by month (%)")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

# Prepare data (uses the "price" you already loaded)
p = price.asfreq("B").ffill()
log_p = np.log(p)

# Build model and predict the next 30 business days
model = ARIMA(log_p, order=(1, 1, 1)).fit()
fc = model.get_forecast(30)

future_dates = pd.bdate_range(p.index[-1] + pd.offsets.BDay(1), periods=30)
forecast = pd.Series(np.exp(fc.predicted_mean.values), index=future_dates)
ci = np.exp(fc.conf_int(alpha=0.05).values)

# Plot
plt.figure(figsize=(12, 5))
plt.plot(p[-250:], label="Past price")
plt.plot(forecast, "r", label="Forecast")
plt.fill_between(future_dates, ci[:, 0], ci[:, 1], color="red", alpha=0.2, label="95% range")
plt.title("30-Day Forecast"); plt.legend(); plt.show()

print("Last price:", round(p.iloc[-1], 2))
print("Forecast in 30 days:", round(forecast.iloc[-1], 2))
print("Likely range:", round(ci[-1, 0], 2), "to", round(ci[-1, 1], 2))
```
